### Conceptos clave que debes conocer
Para que la interfaz y tu código "hablen", necesitas entender estos tres pilares:
1. Signals (Señales) y Slots (Ranuras)
Es el método principal de comunicación en Qt. 

Signal: Un evento que ocurre en la interfaz (ej. "el usuario hizo clic en el botón").
Slot: Una función de Python que se ejecuta cuando recibe esa señal (ej. def mi_funcion():).
Método: Usas .connect() para unirlos: boton.clicked.connect(mi_funcion). 

2. Recibir datos (Input)
Para obtener información del usuario, usarás widgets de entrada:
QLineEdit: Para textos cortos (nombres, rutas de archivos).
QTextEdit: Para párrafos largos.
Método clave: .text() para leer el contenido. 

3. Devolver datos (Output)
Para mostrar resultados de tu código en la interfaz:
QLabel: Para mostrar texto estático o dinámico.
QMessageBox: Para ventanas emergentes de alerta o confirmación.
Método clave: .setText("Nuevo Dato") para actualizar lo que el usuario ve.

#### Guía Rápida de Widgets y Métodos en PySide6

Para que tu código y la interfaz se comuniquen, necesitas conocer los Widgets (elementos visuales) y sus Métodos (funciones para mover datos).

Diccionario de Widgets y Funciones Clave

| Categoría	  | Widget (Clase) | ¿Qué hace?                         | Función para LEER dato      | Función para ESCRIBIR dato |
| :--------   | :------------: | :--------:                         | :--------------------:      | -------------------------: |
| Texto	      | QLabel         | Muestra texto estático o imagenes. | .text()                     | .setText("hola")           |
| Entrada     |	QLineEdit      | Caja para escribir una línea.      | .text()                     |	.setText("ejemplo")        |
| Botón       |	QPushButton    | El clásico botón de clic.          | .isChecked() (si es toggle) | .setText("Click aquí")     |
| Números     |	QSpinBox       | Selector de números (flechitas).   | .value()                    |	.setValue(10)              |
| Opciones    | QComboBox      | Menú desplegable (Lista).          | .currentText()              |	.addItem("Opción 1")       |
| Booleano    |	QCheckBox      | Casilla de verificación.           | .isChecked()                |	.setChecked(True)          |
| Texto Largo |	QTextEdit      | Cuadro para párrafos.              | .toPlainText()              |	.setPlainText("...")       |


#### ¿Cómo se vinculan a tu código? (Eventos)

Para que algo pase cuando el usuario interactúa, cada Widget tiene "Señales" específicas. Estas son las más importantes que debes conocer para empezar:

1. **clicked (QPushButton):** Se activa al hacer clic.
Uso: boton.clicked.connect(tu_funcion)
2. **textChanged (QLineEdit):** Se activa cada vez que el usuario escribe una letra.
Uso: Útil para búsquedas en tiempo real.
3. **valueChanged (QSpinBox):** Se activa cuando cambias el número.
4. **currentIndexChanged (QComboBox):** Se activa cuando el usuario elige otra opción de la lista.



```python
# Creemos nuestra primera aplicacion vacia
from PySide6.QtWidgets import QApplication, QWidget
import sys
# Creamos una instancia de aplicacion
app = QApplication([]) # Tambien sirve sys.argv
# Creamos un QWidget, para nuestra ventana
window = QWidget()
window.show()   # Si no lo ponemos seguira oculta
# Comenzamos el ciclo de eventos
app.exec()
```

```python
# Creamos una interfaz con un unico boton
import sys
from PySide6.QtWidgets import QApplication, QPushButton

app = QApplication(sys.argv)

window = QPushButton("Push Me")
window.show()

app.exec()
```

```python
import sys
from PySide6.QtWidgets import QApplication, QMainWindow

app = QApplication(sys.argv)

window = QMainWindow()
window.show()

# Start the event loop.
app.exec()
```

```python
import sys

from PySide6.QtCore import QSize, Qt
from PySide6.QtWidgets import QApplication, QMainWindow, QPushButton


# Subclass QMainWindow to customize your application's main window
class MainWindow(QMainWindow):
    def __init__(self):
        super().__init__()

        self.setWindowTitle("My App")

        button = QPushButton("Press Me!")

        # Set the central widget of the Window.
        self.setCentralWidget(button)


app = QApplication(sys.argv)

window = MainWindow()
window.show()

app.exec()
```
